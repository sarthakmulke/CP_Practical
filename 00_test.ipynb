```python
#create a simple if elif else code to checkmarks and give
#grade A , B, C, D, F based on the marks entered by the user.
marks = float(input("Enter your marks: "))

if marks >= 80:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 50:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "F"

print("Your grade is: " + grade)
```
